```python
import random

def chatbot():
    print("Chatbot: Hello! Ask me about symptoms.")
    
    while True:
        user_input = input("You: ").lower()

        if "fever" in user_input:
            print("Chatbot: You may have an infection. Please rest and stay hydrated.")
        
        elif "headache" in user_input:
            print("Chatbot: It may be due to stress or dehydration.")
        
        elif "cold" in user_input:
            print("Chatbot: It could be a common cold. Drink warm fluids.")
        
        elif "bye" in user_input:
            print("Chatbot: Take care! Goodbye!")
            break
        
        else:
            print("Chatbot: Sorry, I didn't understand.")

chatbot()
```
